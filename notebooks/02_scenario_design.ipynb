### utility functions

```python
candidate_utility = {

    "Job Description": {
        "Project Manager": 3,
        "Team Manager": 2,
        "Programmer": 1
    },

    "Salary": {
        "120,000 USD": 3,
        "100,000 USD": 2,
        "90,000 USD": 1,
        "60,000 USD": 0
    },

    "Working Hours": {
        "8 hours": 2,
        "9 hours": 1,
        "10 hours": 0
    },

    "Pension Fund": {
        "20%": 2,
        "15%": 1,
        "10%": 0
    },

    "Leased Car": {
        "With leased car": 1,
        "No leased car": 0
    },

    "Promotion Possibilities": {
        "Fast promotion track": 1,
        "Standard promotion": 0,
        "Conditional promotion": 0
    }
}
```

```python
employer_utility = {

    "Job Description": {
        "Programmer": 3,
        "Team Manager": 2,
        "Project Manager": 0
    },

    "Salary": {
        "60,000 USD": 3,
        "75,000 USD": 2,
        "90,000 USD": 1,
        "120,000 USD": 0
    },

    "Working Hours": {
        "10 hours": 2,
        "9 hours": 1,
        "8 hours": 0
    },

    "Pension Fund": {
        "10%": 2,
        "15%": 1,
        "20%": 0
    },

    "Leased Car": {
        "No leased car": 1,
        "With leased car": 0
    },

    "Promotion Possibilities": {
        "Standard promotion": 1,
        "Conditional promotion": 1,
        "Fast promotion track": 0
    }
}
```

```python
def calculate_utility(agreement, utility_table):

    total_utility = 0

    for issue, value in agreement.items():

        if issue in utility_table:

            issue_utilities = utility_table[issue]

            total_utility += issue_utilities.get(value, 0)

    return total_utility
```

```python
##little test 

sample_agreement = {

    "Job Description": "Team Manager",

    "Salary": "90,000 USD",

    "Working Hours": "9 hours",

    "Pension Fund": "10%",

    "Leased Car": "No leased car",

    "Promotion Possibilities": "Fast promotion track"
}

candidate_score = calculate_utility(
    sample_agreement,
    candidate_utility
)

employer_score = calculate_utility(
    sample_agreement,
    employer_utility
)

print("Candidate utility:", candidate_score)
print("Employer utility:", employer_score)
```

```python
utility_tables = {
    "candidate_utility": candidate_utility,
    "employer_utility": employer_utility
}

utility_file = os.path.join(
    scenarios_path,
    "utility_tables.json"
)

with open(utility_file, "w", encoding="utf-8") as f:
    json.dump(utility_tables, f, indent=4)

```
